# 1. NumPy `ndarray`: a multidimensional array object
`ndarray` is an array that can have an arbirary number of dimensions. We will see examples in 2D and 3D for calculations.

**Its main feature: vector calculations are extremely fast!**

Let's see how much faster an `ndarray` object is compared to a loop (with comprehension).
- The magic command `%time` measures the duration of a single command.
- The magic command `%timeit` measures the duration of a single command, running it multiple times in order to provide an error.
- The magic command `%%time` (which we will use in other notebooks) measures the duration of a block of  commands.

````python
import numpy as np
import math as m

nmax = 100000

# For calculations with a comprehension on a list
my_list = list(range(1,nmax))

# For direct calculations on a numpy array
my_arr = np.arange(1,nmax)

n_test = 100

%time for _ in range(n_test): my_list2 = [x*2 for x in my_list]

%time for _ in range(n_test): my_arr2 = my_arr * 2

print(type(my_list), type(my_list2))
print(type(my_arr), type(my_arr2))
````

- `user`: The cumulative time spent by all the CPUs during the computation
- `sys`: The cumulative time spent by all the CPUs during system-related tasks such as memory allocation.
- `Wall time`: The actual time spent in running the process from start to finish, as if it was measured by a human looking at a clock on a wall.

__NumPy based algorithms are generally 10 to 100 times faster than pure Python counterparts!__

Here is another measurement of speedup: it comes with an error bar on the wall-clock time, by running the same command more than once.

````python
%timeit for _ in range(n_test): my_list2 = [x*2 for x in my_list]

%timeit for _ in range(n_test): my_arr2 = my_arr * 2
````

In the examples below you should also notice how much `**` costs in terms of runtime.  **Never forget to try to optimize small things like these: they can make a big difference.**

````python
%timeit for _ in range(n_test): my_list3 = [x**3 for x in my_list]
%timeit for _ in range(n_test): my_list3 = [x*x*x for x in my_list]

%timeit for _ in range(n_test): my_arr3 = my_arr **3
%timeit for _ in range(n_test): my_arr3 = my_arr * my_arr * my_arr
````

````python
%timeit for _ in range(n_test): my_list3 = [m.log(x**2) for x in my_list]
%timeit for _ in range(n_test): my_list3 = [m.log(x*x) for x in my_list]

%timeit for _ in range(n_test): my_arr3 = np.log(my_arr**2)
%timeit for _ in range(n_test): my_arr3 = np.log(my_arr*my_arr)
````

### Applying functions to `ndarray`s

This requires using functions that accept NumPy `ndarray`s as arguments.  E.g., `np.sin` and not `math.sin`

````python
data = np.arange(1,10)

datasq = data**2 # This is the power operator for NumPy `ndarray`s

data_2 = np.log(datasq) + np.sin(data*np.pi)
````

````python
print(data)
print(datasq)
print(data_2)
````

The following fails because the `math` module does not know what to do with the `ndarray` passed as argument of `sqrt`.

````python
m.sqrt(datasq)
````

There is a possible workaround (see next line), but with `ndarray`s you should always use NumPy functions for the runtime benefits they give you on larger collections of numbers.

````python
import math as m
````

````python
tmp = [m.sqrt(val) for val in datasq]
print(tmp)
````

````python
tmp = np.arange(1,10000)

%timeit [m.sqrt(val) for val in tmp]
%timeit np.sqrt(tmp)
````

# 4. Vectorizing functions

To tackle the bottleneck of `for` loops, NumPy provides a vectorization functionality that maps a function over a sequence efficiently.

Let's define a simple example function `pyfunc`.

````python
import numpy as np
````

````python
def pyfunc(a, b, c, d):
    if c*a >= b:
        return a + b + d
    else:
        return a - b - d

# Checking it works
a = 1
b = 2
c = 3
d = 4

pyfunc(a, b, c, d)
````

The interpreter does not know how to handle `pyfunc` when it its first argument is a NumPy array.

````python
a = np.arange(5)
pyfunc(a, b, c, d)
````

# To enable this, we need to create a **vectorized version** of `pyfunc`. vecto_function=np.vectorize(funct)
This weaves throught the input creating tuples of length 4 and running `pyfunc` on it.

````python
vpyfunc = np.vectorize(pyfunc)

# Check it works
print(vpyfunc(a, b, c, d))
print(vpyfunc(a, a, c, d))
print(vpyfunc(a, a, a, d))
print(vpyfunc(a, a, a, a))
````

````python
# Notice that the arguments can also be other kinds of collections,
# which is not the case for the original function
print(vpyfunc(list(a), tuple(a), list(a), tuple(a)))
print(pyfunc(list(a), tuple(a), c, d))
````

````python
# Also notice that the unique weaving through the collections has to be possible, which is not the case below. 
print(vpyfunc((1,2), (1,2,3), c, d))
````

We can enforce the output type.

````python
res = vpyfunc(a, b, c, d)
print(type(res[0]))

vpyfunc =  np.vectorize(pyfunc, otypes=[float])
res = vpyfunc(a, b, c, d)
print(type(res[0]))
````

We can also skip vectorizing on specific arguments, e.g., the 2nd and 4th ones (arguments 1 and 3).

````python
a
````

````python
vpyfunc = np.vectorize(pyfunc, otypes=[float], excluded=[1,3])

print(vpyfunc(a, b, a, d))
print(vpyfunc(b, b, c, d))
````

Indeed it fails if the second or the fourth argument is not a single value.

````python
print(vpyfunc(a, a, c, d))
````

````python
print(vpyfunc(a, b, c, a))
````

Finally, notice that
1. calls to NumPy functions are faster than calls to `math` embedded in a vectorized function
1. vectorization wins over comprehensions when the complexity of the operation you are repeating is high

````python
def pyfunc(x):
    return m.log(x*x) - m.sin(x)*m.cos(x)

vpyfunc = np.vectorize(pyfunc)

%timeit np.log(tmp)*np.log(tmp) - np.sin(tmp)*np.cos(tmp)
%timeit vpyfunc(tmp)
````

````python
%timeit [m.log(x)*m.log(x)**(2-m.log(x)) - m.sin(x)*m.cos(x) for x in tmp]

def pyfunc(x):
    return m.log(x)*m.log(x)**(2-m.log(x)) - m.sin(x)*m.cos(x)

vpyfunc = np.vectorize(pyfunc)


%timeit vpyfunc(tmp)

def pyfunc(x):
    return m.log(x)

vpyfunc = np.vectorize(pyfunc)

# Low complexity: comprehension is faster
%timeit [m.log(x) for x in tmp]
%timeit vpyfunc(tmp)
````

### Extracting columns

````python
print(data)
````

````python
data[:,3] # Fetch the element with index 3 for all values of the 1st index
````

````python
data[1:,:] # Fetch all entries for values of the 1st index from 1 onwards
````

Slicing for a specific index that is out of bounds will cause an error.

However slicing out of bound indices using `:` will not: NumPy will return as many entries as it can.

````python
data[20]
````

````python
data[20:]
````

````python
data[20:20]
````

````python
data[:2,3:]
````

````python
data[:2,3:70]
````

**As with the lists, all operations return a reference to the original array, unless you make a copy**.

In the example below, `y` is a newly defined numpy array, but `x` is not

````python
x = data[:2,1:3] #so x will modify data, y will not
y = np.array(data[:2,1:3])
print(x)
print(y)
````

````python
x[0][1] = -5.55 ##x is a copy, y is not
y[0][1] = 500
print(x)
print(y)
print(data)
````

We have modified not just `x`, but also the original data array!  The same is not true for `y`.

The key difference is that in the `y` case we created a new copy of `data` by calling the constructor `numpy.array`.

# 8. Boolean arrays
As with all operations, also logical operations are vectorised.

````python
import numpy as np
````

````python
data = np.random.normal(0, 1., 9).reshape(3,3) # Same as np.random.normal(0, 1., (3,3))
print(data)

res = data > 0
print(res)
````

**Logical arrays can be used to slice and index an array!**

In this case we want the array with just the positive cells.

````python
posdata = data[res]
print(posdata)
print("# positives: ", len(posdata))
````

Since booleans are converted automatically to 0 and 1, you can easily count them by using the `sum` function

````python
print("Sum of boolean: ", res.sum())
````

Note that the original shape is lost

````python
print(type(posdata))
print("Shape: ", np.shape(posdata))
````

**This implies that slicing with a boolean array creates a new array and not a reference to the original array.**

`all()` and `any()` are important methods when dealing with Boolean arrays.
- `all()` checks if all entries are `True`
- `any()` checks that at least on entry is `True`

````python
print(data>0)

print((data>0).all())

print((data>0).any())
````

# 9. I/O of NumPy objects

NumPy provides built-in functions to easily store and read `ndarray`s in binary and text format **without iterating** over each element.
- Use `save()` and `load()` with binary files (`.npy`).
- Use `savetxt()` and `loadtxt()` with textfiles.

### I/O with `ndarray`s: binary and text files

See [Input/Output with numpy](https://numpy.org/doc/stable/reference/routines.io.html) for a general reference.


**Save to file with `save()` and `savetxt()`**

````python
import numpy as np
````

````python
matrix = np.random.randn(10,5)
print(matrix)

fname = 'npdata1'
np.save(fname+'.npy', matrix)
np.savetxt(fname+'.txt', matrix)
````

Naturally, the binary file uses less memory.

````python
!ls -lrth npdata1*
````

**Reading is also simple with `load()` and `loadtxt()`**

````python
vals = np.load(fname+'.npy')

assert (vals==matrix).all()
````

In particular [numpy.loadtxt](https://numpy.org/doc/stable/reference/generated/numpy.loadtxt.html#numpy.loadtxt) is very handy for quick reading of data from text files.

````python
data = np.loadtxt(fname+'.txt')

assert (data==matrix).all()
````

Slice it to extract columns.

````python
x = data[:,0]
y = data[:,1]
z = data[:,2]
t = data[:,3]
E = data[:,4]

print(z)
````

We can also think of this txt file as having data stored per event (line) in columns (variables).  E.g., the following structure
```
#x   #y   #z   #t   #energy
...  ...  ...  ...  ...
...  ...  ...  ...  ...
...  ...  ...  ...  ...
```
NB: if you don't use usecols, loadtxt will read the rows!!!!!
Since this is a typical usecase, with NumPy we can also **read the data directly into arrays per column** without reading in all data and then manipulating it.  `loadtxt` has the following options to do this:
- `usecols=(a,b,c)` or `usecols=d` reads only the column(s) specified
- `unpack=True` transposes data to go in columns

````python
x,z = np.loadtxt(fname+'.txt', unpack=True, usecols=(0,2))

assert (z==data[:,2]).all()
````

Similarly, [nump.savetxt](https://numpy.org/doc/stable/reference/generated/numpy.savetxt.html) is very convenient to **write `ndarrays` by column to textfile** in a formatted fashion.

````python
fname = 'npdata2.txt'
print(x)
np.savetxt(fname, (x, y, z, t, E), delimiter="\t", fmt="%.3f")

fname = 'npdata3.txt' #here li salvo come colonneee!!
print(x)
np.savetxt(fname, np.array([x, y, z, t, E]).T, delimiter="\t", fmt="%.3f")
````
