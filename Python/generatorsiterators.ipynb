```python
import matplotlib.pyplot as plt
M=int(1e3)
def fibonacci():
    a, b=0, 1
    while True:
     yield a
     a, b = b, a+b
series=fibonacci()
def list_fibonacci(M):
    numb=[0, 1]
    for i in range(M):
      numb.append(numb[-1]+numb[-2])
    return numb

plt.xscale('log')
plt.yscale('log')
plt.plot([next(series) for i in range(M)])
plt.plot(list_fibonacci(M))
plt.show()
```
