```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.fft as ff
```

```python
class Mydataset():
    def __init__(self, N: int, sigma: float =1.):
        np.random.seed(10)
        self.N=N
        self.d=np.random.normal(loc=0., scale=sigma, size=N)
    def DFT(self,freq=True):
        self.D=ff.fft(self.d)
        self.xf = ff.fftfreq(self.N, d=1) 
        if freq:
            return self.D, self.xf
        if not freq:
            return self.D

    def plotDFT(self):
        _ ,_= self.DFT(freq=True)
        fig, axes=plt.subplots(1, 3, figsize=(15, 5))
        
        ax=axes[0]
        ax.set_title("Module of DFT")
        ax.set_xlabel("Frequency")
        ax.set_ylabel("|DFT|")

        ax.plot(self.xf, np.abs(self.D), '.', label="Values", color='orange')
        ax.grid(), ax.legend()
        
        ax=axes[1]
        ax.set_title("Real(DFT)")
        ax.set_ylabel(r"$Re(D_k)$")
        ax.set_xlabel("Frequency")

        ax.plot(self.xf, self.D.real, '.',  label="Values",color='blue')
        ax.grid(), ax.legend()

        ax=axes[2]
        ax.set_title("Im(DFT)")
        ax.set_ylabel(r"$Im(D_k)$")
        ax.set_xlabel("Frequency")

        ax.plot(self.xf, self.D.imag,'.',  label="Values", color='red')
        ax.grid(), ax.legend()


        plt.show()




    def plot(self, nbins=30):
        fig, axes=plt.subplots(1, 2, figsize=(10, 5))
        
        ax=axes[0]
        ax.set_title("Data ordered by index")
        ax.set_xlabel("index")
        ax.set_ylabel("Value extracted")

        ax.plot(self.d, '.', label="Values", color='orange')
        ax.grid(), ax.legend()
        
        ax=axes[1]
        ax.set_title("Data histogram")
        ax.set_ylabel("Counts")
        ax.set_xlabel("Value extracted")

        ax.hist(self.d, label="Values", bins=nbins,color='blue')
        ax.grid(), ax.legend()
        plt.show()
```

```python
d=Mydataset(100)
d.plot()
a, b =d.DFT()
d.plotDFT()
a, b, a.imag, a.real
```
