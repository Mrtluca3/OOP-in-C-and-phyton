```python
import numpy as np
```

```python
class Solid:
    def __init__(self, M, l1, InertiaCM, l2=0., l3=0.):
        self.M=M
        self.lengths=np.array([l1, l2, l3])
        self.omega=0.
        self.InertiaCM=InertiaCM
        self.Inertia=InertiaCM
        self.d=0.
        self.K=0.
    
    def spin(self, Omega,  d=0.):
        self.d=d
        self.omega=Omega
        self.Inertia=self.InertiaCM+ self.M*d*d
        self.K=0.5 * (self.Inertia) * Omega*Omega
        return self
```

```python
class Sphere(Solid):
    def __init__(self, M, Radius):
        InertiaCM=2./5. * M * Radius * Radius
        super().__init__(M, Radius, InertiaCM)
        self.params={
            'Radius': self.lengths[0],
            'InertiaCM': self.InertiaCM,
            'Inertia': self.Inertia,
            'Mass': self.M,
            'd': self.d,
            'omega': self.omega,
            'K': self.K
            }
        

    def get(self, *args):
        # return self.params[args]
        return [self.params[arg] for arg in args if arg in self.params]
    def set(self, **kargs):        
        for key, value in kargs.items():
            if key in self.params:
                self.params[key]=value
        return self
    def print(self, text=""):
        print(text)
        for key, value in self.params.items():
            print(key, " ", value)
        print("")
        return self

s=Sphere(10, 2)
s.get('Mass', 'd', 'Radius', 'K')
s.print()
s.set(Mass=100)
s.print()
s.spin(7, d=0.1)
s.print("After spin")

```
